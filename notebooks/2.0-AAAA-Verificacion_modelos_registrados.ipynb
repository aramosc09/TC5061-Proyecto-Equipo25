```python
import mlflow
```

```python
# 1. Conectar al servidor donde está el Model Registry
mlflow.set_tracking_uri("http://127.0.0.1:5000")
```

```python
uri_modelo_p1 = "models:/mibici-equipo25-p1-demanda@champion"

# Extraer metadatos (Firma y Run ID)
info_modelo_p1 = mlflow.models.get_model_info(uri_modelo_p1)

print("   * Firma del Modelo:")
print(info_modelo_p1.signature)

print("   * Run ID de Origen:")
print(info_modelo_p1.run_id)

# Cargar el modelo original usando la variante de sklearn
# Devuelve el pipeline o clase original directamente, sin envoltorios.
modelo_sklearn_p1 = mlflow.sklearn.load_model(uri_modelo_p1)

print("   * Parámetros del Algoritmo Interno:")
# Como ya es el objeto original, puedes usar sus métodos nativos directamente:
print(modelo_sklearn_p1.get_params())
```
